```python
import os
import dotenv
import pandas as pd
from kaggle.api.kaggle_api_extended import KaggleApi

dotenv.load_dotenv()
```

```python
os.environ['KAGGLE_USERNAME'] = os.getenv("KAGGLE_USERNAME")
os.environ['KAGGLE_KEY'] = os.getenv("KAGGLE_KEY")

api = KaggleApi()
api.authenticate()
```

```python
def download_dataset(dataset, path):
    api.dataset_download_files(dataset, path=path, unzip=True)
```

```python
dataset = os.getenv("DATASET")
path = f'../data/kaggle/raw/{dataset.split("/")[1]}/'

download_dataset(dataset, path)
```

```python
csvs = os.listdir(path)
csvs
```

```python
df = pd.read_csv(f'{path}/{csvs[3]}', sep=';')
df = df.dropna()
```

```python
df
```

```python
clean_df = df[[
    'age',
    'job',
    'marital',
    'education',
    'balance',
    'housing',
    'loan',
    'pdays',
    'previous',
    'poutcome'
]]
clean_df
```

```python
clean_dataset_path = str(os.getenv("CLEAN_DATASET_PATH"))

os.makedirs(clean_dataset_path, exist_ok=True)
clean_df.to_parquet(f'{clean_dataset_path}/clean_bank.parquet', index=False)
```
